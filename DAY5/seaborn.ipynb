```python
# -----------------------------------
#  Introduction to Seaborn
# Seaborn is a Python data visualization library built on top of Matplotlib.
# It provides a high-level interface for drawing attractive and informative statistical graphics.
# -----------------------------------
#pip install seaborn
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Load sample dataset from Seaborn
tips = sns.load_dataset("tips")  # dataset of restaurant bills
print(tips.head())
```

```python
# -----------------------------------
#  Scatter Plot
# Used to visualize the relationship between two numerical variables
# Example: total_bill vs tip
# -----------------------------------

sns.scatterplot(x="total_bill", y="tip", data=tips)
plt.title("Scatter Plot: Total Bill vs Tip")
plt.show()
```

```python
# -----------------------------------
#  Regression Plot
# Shows scatter plot + best fit line (trend line)
# Helps to see linear relationship
# -----------------------------------

sns.regplot(x="total_bill", y="tip", data=tips)
plt.title("Regression Plot: Total Bill vs Tip")
plt.show()
```

```python
# -----------------------------------
#  Bar Plot
# Used to compare categorical data with numerical values
# Example: Average tip given by Male vs Female
# -----------------------------------

sns.barplot(x="sex", y="tip", data=tips)
plt.title("Bar Plot: Average Tip by Gender")
plt.show()
```

```python
# -----------------------------------
#  Count Plot
# Shows the count of each category (frequency plot)
# Example: Number of Male vs Female customers
# -----------------------------------

sns.countplot(x="sex", data=tips)
plt.title("Count Plot: Number of Male vs Female Customers")
plt.show()
```

```python
# -----------------------------------
#  Box Plot (Whisker Plot)
# Used to visualize distribution and outliers
# Example: Tip distribution by Gender
# -----------------------------------

sns.boxplot(x="sex", y="tip", data=tips)
plt.title("Box Plot: Tip Distribution by Gender")
plt.show()
```

```python
# -----------------------------------
#  Violin Plot
# Combination of box plot + KDE (density plot)
# Shows distribution + probability density
# Example: Tip distribution by Day
# -----------------------------------

sns.violinplot(x="day", y="tip", data=tips)
plt.title("Violin Plot: Tip Distribution by Day")
plt.show()
```

```python
# -----------------------------------
#  Heatmap
# Used to visualize correlation between variables
# Darker/lighter colors represent strength of relationship
# -----------------------------------

corr = tips.corr(numeric_only=True)  # calculate correlation matrix
sns.heatmap(corr, annot=True, cmap="coolwarm", linewidths=0.5)
plt.title("Heatmap: Correlation between Numerical Features")
plt.show()
```

```python

import numpy as np
from mpl_toolkits.mplot3d import Axes3D   # 3D plots ke liye

# Sample dataset
tips = sns.load_dataset("tips")
```

```python
# -----------------------------------
# Pairplot
# Automatically plots relationship between multiple numerical columns
# Scatterplots + Histograms together
# -----------------------------------
sns.pairplot(tips, hue="sex")  # color by gender
plt.show()
```

```python
# -----------------------------------
#  Jointplot
# Shows scatter plot + histograms + regression line
# Helps to analyze relation between two variables with distributions
# -----------------------------------
sns.jointplot(x="total_bill", y="tip", data=tips, kind="reg", color="green")
plt.show()
```

```python
# -----------------------------------
#  KDE Plot
# Kernel Density Estimation = Smoothed curve of distribution
# Better than histogram for continuous data
# -----------------------------------
sns.kdeplot(data=tips["total_bill"], fill=True, color="purple")
plt.title("KDE Plot: Distribution of Total Bill")
plt.show()
```

```python
# -----------------------------------
#  Stripplot & Swarmplot
# Stripplot -> individual data points
# Swarmplot -> non-overlapping distribution of points
# -----------------------------------
sns.stripplot(x="day", y="tip", data=tips, jitter=True, color="orange")
plt.title("Strip Plot: Tips by Day")
plt.show()

sns.swarmplot(x="day", y="tip", data=tips, color="blue")
plt.title("Swarm Plot: Tips by Day")
plt.show()
```

```python
# -----------------------------------
#  3D Scatter Plot
# Used to visualize relation between 3 numerical variables
# -----------------------------------
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

x = tips["total_bill"]
y = tips["tip"]
z = tips["size"]

ax.scatter(x, y, z, c=z, cmap="viridis", marker="o")

ax.set_xlabel("Total Bill")
ax.set_ylabel("Tip")
ax.set_zlabel("Table Size")
ax.set_title("3D Scatter Plot")
plt.show()
```

```python
# -----------------------------------
#  3D Line Plot
# Useful for time-series or sequence data
# -----------------------------------
t = np.linspace(0, 20, 100)
x = np.sin(t)
y = np.cos(t)
z = t

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot(x, y, z, color="red", label="3D Line")

ax.set_title("3D Line Plot")
ax.legend()
plt.show()
```

```python
# -----------------------------------
#  3D Surface Plot
# Shows continuous surface in 3D space (like hills and valleys)
# -----------------------------------
x = np.linspace(-5, 5, 50)
y = np.linspace(-5, 5, 50)
x, y = np.meshgrid(x, y)
z = np.sin(np.sqrt(x**2 + y**2))

fig = plt.figure()
ax = fig.add_subplot(111, projection="3d")
surf = ax.plot_surface(x, y, z, cmap="coolwarm")

fig.colorbar(surf)
ax.set_title("3D Surface Plot")
plt.show()
```

```python
'''Scatterplot → Relation between 2 numbers

Regression Plot → Scatter + Trend line

Bar Plot → Categorical vs Numerical

Count Plot → Frequency of categories

Box Plot → Spread + Outliers

Violin Plot → Box + Density curve

Heatmap → Correlation matrix'''
```

```python
'''Seaborn → Easy & attractive (pairplot, jointplot, heatmap, violin, etc.)

Matplotlib → Powerful & customizable (3D scatter, line, surface)

Dynamic Charts → Pairplot (all in one), Jointplot, KDE, Swarmplot

3D Plots → Scatter, Line, Surface'''
```
