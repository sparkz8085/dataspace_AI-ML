```python
# ----------------------------------------
#  Matplotlib Introduction
# ----------------------------------------

# Matplotlib is a Python library used for data visualization.
# It helps to create static, animated, and interactive plots.
# It is especially useful in Data Analysis and Machine Learning
# to represent data graphically.

import matplotlib.pyplot as plt
import numpy as np
```

```python
# Create data
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

# Plot the data
plt.plot(x, y)

# Add labels and title
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.title("Basic Line Plot")

# Show the plot
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [1, 4, 9, 16, 25]

# Different styles
plt.plot(x, y, color="red", linestyle="--", marker="o")

plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.title("Line with Style, Color, and Markers")
plt.show()
```

```python
x = [1, 2, 3, 4, 5]

plt.plot(x, [i for i in x], label="y=x")
plt.plot(x, [i**2 for i in x], label="y=x^2")
plt.plot(x, [i**3 for i in x], label="y=x^3") 
    

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Multiple Lines")
plt.legend()   # Show legend
plt.show()
```

```python
# Categories
subjects = ["Math", "Science", "English", "History"]
marks = [90, 80, 75, 85]

plt.bar(subjects, marks, color="red")

plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.title("Bar Chart Example")
plt.show()
```

```python
plt.barh(subjects, marks, color="green")
plt.xlabel("Marks")
plt.ylabel("Subjects")
plt.title("Horizontal Bar Chart")
plt.show()
```

```python
# Generate random data
data = np.random.randn(1000)

plt.hist(data, bins=30, color="orange", edgecolor="black")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram Example")
plt.show()
```

```python
x = np.random.rand(50)
y = np.random.rand(50)

plt.scatter(x, y, color="red")

plt.xlabel("X Random")
plt.ylabel("Y Random")
plt.title("Scatter Plot Example")
plt.show()
```

```python
sizes = [20, 30, 25, 25]
labels = ["A", "B", "C", "D"]
colors = ["gold", "lightblue", "lightgreen", "pink"]

plt.pie(sizes, labels=labels, colors=colors, autopct="%1.1f%%")

plt.title("Pie Chart Example")
plt.show()
```

```python
# Data
sizes = [30, 20, 25, 25]
labels = ["A", "B", "C", "D"]

# 1️ No autopct (just labels)
plt.subplot(2,2,1)
plt.pie(sizes, labels=labels)
plt.title("No autopct")

# 2️ Integer percentage (no decimal)
plt.subplot(2,2,2)
plt.pie(sizes, labels=labels, autopct="%d%%")
plt.title("autopct = %d%%")

# 3️ One decimal place
plt.subplot(2,2,3)
plt.pie(sizes, labels=labels, autopct="%0.1f%%")
plt.title("autopct = %0.1f%%")

# 4️ Two decimal places
plt.subplot(2,2,4)
plt.pie(sizes, labels=labels, autopct="%0.2f%%")
plt.title("autopct = %0.2f%%")

plt.tight_layout()
plt.show()
```

```python
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

# Create 1 row and 2 columns of plots
plt.subplot(1, 2, 1)
plt.plot(x, y1, color="blue")
plt.title("Sine Function")

plt.subplot(1, 2, 2)
plt.plot(x, y2, color="red")
plt.title("Cosine Function")

plt.show()
```

```python
plt.figure(figsize=(6,4))   # Change figure size
plt.plot(x, y1, color="purple")
plt.title("Custom Size Plot")
plt.savefig("myplot.png")   # Save as image
plt.show()
```

```python


subjects = ["Math", "Science", "English", "History"]
classA = [90, 80, 75, 85]
classB = [70, 60, 65, 78]

# First draw Class A bars
plt.bar(subjects, classA, label="Class A", color="skyblue")

# Then draw Class B on top of Class A
plt.bar(subjects, classB, bottom=classA, label="Class B", color="orange")

plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.title("Stacked Bar Chart")
plt.legend()
plt.show()
```

```python


subjects = ["Math", "Science", "English", "History"]
classA = [90, 80, 75, 85]
classB = [70, 60, 65, 78]

x = np.arange(len(subjects))   # positions [0,1,2,3]
width = 0.35                   # bar width

# Class A bars shifted left
plt.bar(x - width/2, classA, width, label="Class A", color="blue")

# Class B bars shifted right
plt.bar(x + width/2, classB, width, label="Class B", color="green")

plt.xticks(x, subjects)   # Replace index with subject names
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.title("Grouped Bar Chart (Class A vs Class B)")
plt.legend()
plt.show()
```

```python
sizes = [40, 25, 20, 15]
labels = ["A", "B", "C", "D"]
colors = ["gold", "lightblue", "lightgreen", "pink"]

plt.pie(
    sizes,
    labels=labels,
    colors=colors,
    autopct="%1.1f%%",   # show percentage with 1 decimal
    startangle=90,       # rotate chart so first slice starts at 90 degrees
    explode=(0.1, 0, 0, 0), # pull out first slice slightly
    shadow=True          # add shadow for 3D effect
)

plt.title("Customized Pie Chart")
plt.show()
```

```python


# Generate 1000 random numbers (normal distribution)
data = np.random.randn(1000)

plt.hist(
    data,
    bins=20,         # number of bars (default ~10)
    color="purple",
    edgecolor="black",
    density=True     # normalize so area = 1 (probability density)
)

plt.xlabel("Value")
plt.ylabel("Frequency / Density")
plt.title("Histogram Example")
plt.show()
```

```python

from mpl_toolkits.mplot3d import Axes3D


# Create data
z = np.linspace(0, 20, 100)    # z-axis
x = np.sin(z)                  # x = sin(z)
y = np.cos(z)                  # y = cos(z)

# Create figure
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Plot 3D line
ax.plot(x, y, z, label="3D Line")
ax.set_xlabel("X Axis")
ax.set_ylabel("Y Axis")
ax.set_zlabel("Z Axis")
ax.set_title("3D Line Plot")
ax.legend()

plt.show()
```

```python
# Random data
x = np.random.rand(50)
y = np.random.rand(50)
z = np.random.rand(50)

fig = plt.figure()
ax = fig.add_subplot(111, projection="3d")

ax.scatter(x, y, z, c="red", marker="o")

ax.set_xlabel("X Axis")
ax.set_ylabel("Y Axis")
ax.set_zlabel("Z Axis")
ax.set_title("3D Scatter Plot")

plt.show()
```

```python
#from mpl_toolkits.mplot3d import Axes3D
from matplotlib import cm   # colormap

# Meshgrid data
x = np.linspace(-5, 5, 50)
y = np.linspace(-5, 5, 50)
x, y = np.meshgrid(x, y)
z = np.sin(np.sqrt(x**2 + y**2))

fig = plt.figure()
ax = fig.add_subplot(111, projection="3d")

# Surface plot
surf = ax.plot_surface(x, y, z, cmap=cm.viridis)

ax.set_xlabel("X Axis")
ax.set_ylabel("Y Axis")
ax.set_zlabel("Z Axis")
ax.set_title("3D Surface Plot")

plt.colorbar(surf)  # color scale
plt.show()
```

```python
x = [1,2,3,4,5]
y = [2,3,4,5,6]
z = [0,0,0,0,0]   # starting height

dx = dy = [0.5,0.5,0.5,0.5,0.5]  # bar width
dz = [1,2,3,4,5]  # bar heights

fig = plt.figure()
ax = fig.add_subplot(111, projection="3d")

ax.bar3d(x, y, z, dx, dy, dz, color="cyan")

ax.set_xlabel("X Axis")
ax.set_ylabel("Y Axis")
ax.set_zlabel("Z Axis")
ax.set_title("3D Bar Plot")

plt.show()
```
